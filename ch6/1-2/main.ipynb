## Random Forests

random forest is an ensemble of decision trees который обучен через
bagging method (or sometimes pasting) где max_samples = size of the training set
Вместо создания BaggingClassifier и вставки DecisionTreeClassifier можно 
просто использовать RandomForestClassiefier который более удобный и уже
оптимизированный под decinion trees

```python

from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X, y = make_moons(n_samples=500, noise=0.30, random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

rnd_clf = RandomForestClassifier(n_estimators=500, max_leaf_nodes=16,
n_jobs=-1, random_state=42)
rnd_clf.fit(X_train, y_train)
y_pred_rf = rnd_clf.predict(X_test)
print(rnd_clf.score(X_test, y_test))
```

## Feature Importance

```python
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier 
iris = load_iris(as_frame=True)
rnd_clf = RandomForestClassifier(n_estimators=500, random_state=42)
rnd_clf.fit(iris.data, iris.target)

# То есть по каждой колонке говорит на сколько она важна
print(rnd_clf.feature_importances_)
print(iris.data.columns)
print(iris.target_names)
# Более важные тут оказались petal length и petal width
for score, name in zip(rnd_clf.feature_importances_, iris.data.columns):
    print(round(score, 2), name)

# В сумме если сложить feature_importances_ они дают 1
# Вывод: решают лепестки (44% + 42% = 86%), чашелистики почти не влияют.
```
