# Таблица 6-1. Когда использовать каждый метод ансамблей

| Метод ансамбля | Когда использовать | Примеры задач |
|---|---|---|
| **Hard Voting** (жёсткое голосование) | Сбалансированный датасет классификации с несколькими сильными, но разнообразными классификаторами. | Определение спама, анализ тональности, классификация заболеваний |
| **Soft Voting** (мягкое голосование) | Датасет классификации с вероятностными моделями, где важны оценки уверенности. | Медицинская диагностика, оценка кредитного риска, детекция фейковых новостей |
| **Bagging** (бэггинг) | Структурированные или полуструктурированные данные с высокой дисперсией и моделями, склонными к переобучению. | Финансовое моделирование рисков, рекомендации в e-commerce |
| **Pasting** (пастинг) | Структурированные или полуструктурированные данные, где нужны более независимые модели. | Сегментация клиентов, классификация белков |
| **Random Forest** (случайный лес) | Высокоразмерные структурированные данные с потенциально шумными признаками. | Прогноз оттока клиентов, анализ генетических данных, детекция мошенничества |
| **Extra-Trees** (экстремально случайные деревья) | Большие структурированные датасеты со множеством признаков, где важна скорость и снижение дисперсии. | Детекция мошенничества в реальном времени, анализ данных сенсоров |
| **AdaBoost** | Малые и средние структурированные датасеты с низким шумом и слабыми учениками (например, пеньками), где полезна интерпретируемость. | Кредитный скоринг, детекция аномалий, предиктивное обслуживание |
| **Gradient Boosting** (градиентный бустинг) | Средние и большие структурированные датасеты, где требуется высокая предсказательная сила, даже ценой дополнительной настройки. | Прогноз цен на жильё, оценка рисков, прогнозирование спроса |
| **Histogram-Based Gradient Boosting (HGB)** | Большие структурированные датасеты, где ключевы скорость обучения и масштабируемость. | Прогноз CTR (click-through rate), ранжирование, real-time bidding в рекламе |
| **Stacking** (стекинг) | Сложные высокоразмерные датасеты, где объединение нескольких разнообразных моделей может максимизировать точность. | Рекомендательные системы, принятие решений в автономных автомобилях, соревнования Kaggle |
