```python
# Splitting Train and Test Data
from sklearn.model_selection import train_test_split
import pandas as pd

emails_full = pd.read_csv("./dataset.csv")

X = emails_full.drop(columns=["isSpam"])
y = emails_full["isSpam"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)


X_train.info()
```

```python
# Creating a vectorizer
from sklearn.feature_extraction.text import CountVectorizer
import string


# Todo should be pipeline!

def cleanTexts(texts: list[str]):
		return [
			text.lower().translate(str.maketrans("", "", string.punctuation))
			for text in texts
		]

vectorizer = CountVectorizer()

X_train_vectors = vectorizer.fit_transform(cleanTexts(X_train['body']))
X_test_vectors = vectorizer.transform(cleanTexts(X_test['body']))
print(X_train_vectors.shape)

```

```python
# Creating another vectorizer with stop words
vectorizer_stop = CountVectorizer(stop_words="english")
vectors_stop = vectorizer_stop.fit_transform(cleanTexts(X_train['body']))

print(vectors_stop.shape)
```

```python
# Analyze frequent words

import numpy as np

def analyzeFrequentWords(v):
	# Sum all words, axis=0 означает по строке
	word_counts = np.asarray(v.sum(axis=0)).ravel()
	# argsort вернуть отсортированные ИНДЕКСЫ по значению, [::-1] reverse последовательности 
	top_indices = word_counts.argsort()[-20:][::-1]
	for i in top_indices:
		print(vectorizer.get_feature_names_out()[i], word_counts[i])
```

```python
analyzeFrequentWords(X_train_vectors)
```

```python
analyzeFrequentWords(vectors_stop)
```
