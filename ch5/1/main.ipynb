```python
# Training and Visualizing a Decision Tree

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import export_graphviz


iris = load_iris(as_frame=True)
X_iris = iris.data[["petal length (cm)", "petal width (cm)"]].values
y_iris = iris.target
tree_clf = DecisionTreeClassifier(max_depth=2, random_state=42)
tree_clf.fit(X_iris, y_iris)

# Для визуализации можно использовать export_graphviz
export_graphviz(
    tree_clf,
    out_file="iris_tree.dot",
    feature_names=["petal length (cm)", "petal width (cm)"],
    class_names=iris.target_names,
    rounded=True,
    filled=True,
)
```

```python
# После этого созданный файл можно визуализировать
# Но перед этим сделай установку
# https://graphviz.org/download

from graphviz import Source
Source.from_file("iris_tree.dot") 
```

G_i = 1 - \sum_{k=1}^{n} p_{i,k}^2

p_{i,k} = n_{i,k} / m_i

где n_{i,k} — сколько в узле i примеров класса k, а m_i — сколько всего примеров в узле i.
