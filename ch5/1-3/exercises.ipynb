   - d. Оцените эти предсказания на тестовой выборке: вы должны получить немного более высокую точность, чем у первой модели (примерно на 0,5–1,5% выше). Поздравляем, вы обучили классификатор «случайный лес»!

```python
# 7
# a. Используйте `make_moons(n_samples=10000, noise=0.4)` для генерации датасета moons.
# b. Используйте `train_test_split()` для разделения датасета на обучающую и тестовую выборки.
# c. Используйте решётчатый поиск с кросс-валидацией (с помощью класса `GridSearchCV`), чтобы найти
# хорошие значения гиперпараметров для `DecisionTreeClassifier`. Подсказка: попробуйте различные
# значения `max_leaf_nodes`.*
# d. Обучите модель на всей обучающей выборке с этими гиперпараметрами и измерьте качество модели на
# тестовой выборке. Вы должны получить примерно 85–87% точности.

from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.tree import DecisionTreeClassifier

X_moons, y_moons = make_moons(n_samples=10000, noise=0.4, random_state=42)
x_train_data, x_test_data, y_train_data, y_test_data  = train_test_split(X_moons, y_moons, test_size=0.2, random_state=42)

param_grid = {"max_leaf_nodes": [10, 5, 7, 13, 4, 6, 16, 20, 30]}

model = DecisionTreeClassifier(random_state=42)
grid_search = GridSearchCV(model, param_grid, cv=5, n_jobs=1)
grid_search.fit(x_train_data, y_train_data)
print(f"Best Params: {grid_search.best_params_}")
print(f"Best Score: {grid_search.best_score_}")
print(grid_search.score(x_test_data, y_test_data))
```
