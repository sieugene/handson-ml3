# Elastic Net (Equation 4-12)

Название **elastic** = «эластичный, упругий» — метод «тянется» между Ridge и Lasso, как резинка, и может остановиться в любой точке между ними.

## Формула

```
J(θ) = MSE(θ) + r(2α ∑ᵢ₌₁ⁿ |θᵢ|) + (1 − r)(α/m ∑ᵢ₌₁ⁿ θᵢ²)
```

Это буквально сумма двух штрафов, которые вы уже разбирали:

| Кусок формулы | Откуда он | Что это |
|---|---|---|
| `MSE(θ)` | база | Обычная ошибка предсказания |
| `2α ∑\|θᵢ\|` | из Lasso (Equation 4-10) | ℓ1-штраф — сумма модулей весов |
| `α/m ∑θᵢ²` | из Ridge | ℓ2-штраф — сумма квадратов весов |
| `r` | новое | Коэффициент смешивания (mix ratio), от 0 до 1 |
| `(1 − r)` | новое | «Обратный» коэффициент — противовес для `r` |

## Что делает `r`

`r` — это рубильник между двумя штрафами:

- **r = 1** → `(1 − r) = 0`, ridge-слагаемое обнуляется → остаётся только `r · (2α∑|θᵢ|)` = **чистый Lasso**.
- **r = 0** → сам множитель `r` обнуляет lasso-слагаемое → остаётся только `(1−r) · (α/m ∑θᵢ²)` = **чистый Ridge**.
- **0 < r < 1** → смесь: часть штрафа обнуляет веса (как Lasso), часть просто уменьшает их (как Ridge).

Как код: `r * lasso_penalty + (1 - r) * ridge_penalty` — просто взвешенное среднее двух штрафов.

## Когда что использовать (по книге)

- **Plain linear regression** (без регуляризации) — почти никогда не стоит использовать.
- **Ridge** — хороший вариант по умолчанию.
- **Lasso / Elastic Net** — если подозреваете, что полезны только несколько признаков (остальные веса обнулятся).
- **Elastic Net предпочтительнее Lasso**, потому что Lasso ведёт себя нестабильно, когда признаков больше, чем обучающих примеров, или когда несколько признаков сильно коррелируют друг с другом.

## Scikit-Learn

```python
from sklearn.linear_model import ElasticNet
elastic_net = ElasticNet(alpha=0.1, l1_ratio=0.5)  # l1_ratio = r
elastic_net.fit(X, y)
elastic_net.predict([[1.5]])
```

`l1_ratio` в коде — это и есть `r` из формулы: `l1_ratio=1.0` → Lasso, `l1_ratio=0.0` → Ridge.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import ElasticNet

np.random.seed(42)
m = 20
X = 3 * np.random.rand(m, 1)
y = 1 + 0.5 * X + np.random.randn(m, 1) / 1.5

elastic_net = ElasticNet(alpha=0.1, l1_ratio=0.5)
elastic_net.fit(X, y)
elastic_net.predict([[1.5]])
````
