```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([[0], [0.5], [1], [1.5], [2]])
y = np.array([[4], [5.5], [7], [8.5], [10]])
print("X", x)
print("y", y)
```

```python
import matplotlib.pyplot as plt

plt.plot(x, y, "b.") # b = blue, . = dots
plt.show()
```

```python
plt.plot(x, y, "b.") # b = blue, . = dots
plt.plot([0, 4], [2, 10], "r-")  # r = red, - = line
```

```python
from sklearn.preprocessing import add_dummy_feature
X_b = add_dummy_feature(x)
theta_best = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
print(theta_best)
```

```python
plt.plot(x, y, "b.") # b = blue, . = dots
new_raw = np.array([[0], [4]]) # Это та самая линия
print(new_raw)
# add_dummy_feature добавляет столбец, без него не можем сделать new @ theta_best
# [0]  -->  [[1, 0],
# [4]  -->  [1, 4]]
# Зачем это нужно? 1 ≠ 2 → не умножается.
# У каждой таблицы 2 строки и 1 столбец. количество столбцов слева должно быть равно количеству строк справа.
# Наша theta_best!
# [[0]
# [4]]
new = add_dummy_feature(new_raw)
# plt.plot(new) -> plt.plot(new_raw) поменяй и поймешь
plt.plot(new, new @ theta_best, "r-")  # r = red, - = line
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create 2 rows, 1 column of plots
fig, axs = plt.subplots(2, 1, figsize=(6, 8))

# Plot on the first (top) axes
axs[0].plot(x, y1, color='blue')
axs[0].set_title('Sine Wave')

# Plot on the second (bottom) axes
axs[1].plot(x, y2, color='red')
axs[1].set_title('Cosine Wave')

# Adjust layout so labels do not overlap
plt.tight_layout()
plt.show()
```
