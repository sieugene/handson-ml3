```python
# rng.standard_normal разбор


# нормальное распределение, где:
# среднее = 0
# стандартное отклонение = 1
# значения чаще всего находятся около 0
# отрицательные и положительные значения возможны

print(rng.standard_normal((2, 3)))
```

```python
# Разбор строки y = 0.5 * X**2 + X + 2 + rng.standard_normal((m, 1))
rng = np.random.default_rng(seed=42)

example_x = rng.random((1,2))
std_n = rng.standard_normal((1, 2))
result = 0.5 * example_x**2 + example_x + 2 + std_n
print("result:",result,"\n")

# X**2 поэлементно каждый возводится во вторую степень
x_2_ex = example_x
print("ориг",x_2_ex)
print("возвели", x_2_ex**2,"\n")

# После возведения мы перемножаем на 0.5
x_2_ex_af =0.5 * x_2_ex**2
print("Возведенное перемножаем на 0.5",x_2_ex_af,"\n") 

# Снова складываем, прибавляем себя же, но ориг и каждый складываем 2
x_2_ex_sum_it = x_2_ex_af + x_2_ex + 2
print(x_2_ex_sum_it, "\n")

# Делаем норм рандом распределение
print("std_n:", std_n, "\n")
# И складываем это
print("финальный",x_2_ex_sum_it + std_n)
```
