### Softmax (из книги)

В книге формула записана в полном виде:

$$
\hat{p}_k = \sigma(s(x))_k =
\frac{\exp(s_k(x))}
{\sum_{j=1}^{K}\exp(s_j(x))}
$$

Мы используем сокращённую запись:

$$
\hat{p}_k =
\frac{e^{s_k}}
{\sum_j e^{s_j}}
$$

Это **одна и та же формула**, просто некоторые обозначения были сокращены.

| В книге          | В сокращённой записи | Что означает                             |
| ---------------- | -------------------- | ---------------------------------------- |
| $\exp(s_k(x))$   | $e^{s_k}$            | Экспонента score класса $k$              |
| $\sum_{j=1}^{K}$ | $\sum_j$             | Суммируем по всем классам                |
| $s_k(x)$         | $s_k$                | Score класса $k$ для объекта $x$         |
| $K$              | —                    | Количество классов, понятно из контекста |
| $\sigma(s(x))_k$ | —                    | Результат Softmax для класса $k$         |

То есть:

$$
\exp(s_k(x)) \equiv e^{s_k}
$$

и

$$
\sum_{j=1}^{K}\exp(s_j(x))
\equiv
\sum_j e^{s_j}
$$

Мы просто не указываем $x$ и границы суммы явно, потому что они уже понятны из контекста.

**Важно:** в книге $s_k(x)$ — это **score (logit)** класса $k$, а $\hat{p}_k$ — уже **вероятность** после применения Softmax.


```python
import numpy as np

thetas = np.array([
	# cat
	[2,1],
	# dog
	[1,2],
	# rabbit
	[0.5,1]
])
x = [3,2]

# 1. Считаем score
def getScore():
	scores = []
	for theta in thetas:
		# 2 * 3 + 1 * 2 ...
		scores.append(theta @ x)
	return scores
scores = getScore()
print("scores:", scores)

# 2. Превращаем в вероятности
def softmax():
	_probs = []
	sum_of_exp_scores = np.sum(np.exp(scores))
	for score in scores:
		e_s_k_exp_score = np.exp(score)
		_probs.append(e_s_k_exp_score / sum_of_exp_scores)
	return _probs

print("softmax:",softmax())

# 3. argmax
print("argmax:",np.argmax(softmax()))

```

```python
import numpy as np

# 1) Сырые баллы, которые «выдала» модель (на самом деле их считали
#    по формуле θ^T x, но мы просто берём готовые, чтобы показать softmax)
scores = np.array([4.5, 2.0, 0.1])   # [кот, собака, машина]

# 2) SOFTMAX — та самая формула №2, записанная в 3 строки
def softmax(scores):
    exp_scores = np.exp(scores)          # числитель для каждого класса
    total = exp_scores.sum()             # знаменатель (сумма по всем K классам)
    return exp_scores / total            # делим каждый на общий пирог

probs = softmax(scores)
print(probs)
# [0.91391132 0.0753806  0.01070808]

print("Сумма:", probs.sum())   # 1.0 — ровно 100%, как и обещали

# 3) Какой класс победил?
print("Прогноз:", ["кот", "собака", "машина"][np.argmax(probs)])
# Прогноз: кот
```

```python
# x — признаки объекта (2 признака для простоты)
x = np.array([2.0, 5.0])

# Θ — таблица линеек: одна строка на класс
#   строка 0 — линейка кота
#   строка 1 — линейка собаки
#   строка 2 — линейка машины
Theta = np.array([
    [1.0, 0.5],   # кот
    [0.2, 0.3],   # собака
    [-0.5, 0.1],  # машина
])

# Формула s_k(x) = θ^(k) · x  для ВСЕХ классов сразу:
# x.T — переворот (то самое «T» в формуле), @ — перемножение матриц
scores = Theta @ x
print("Сырые баллы:", scores)
# [ 4.5   1.9  -0.9]

probs = softmax(scores)
print("Вероятности:", probs)
```
