### Softmax (из книги)

В книге формула записана в полном виде:

$$
\hat{p}_k = \sigma(s(x))_k =
\frac{\exp(s_k(x))}
{\sum_{j=1}^{K}\exp(s_j(x))}
$$

Мы используем сокращённую запись:

$$
\hat{p}_k =
\frac{e^{s_k}}
{\sum_j e^{s_j}}
$$

Это **одна и та же формула**, просто некоторые обозначения были сокращены.

| В книге          | В сокращённой записи | Что означает                             |
| ---------------- | -------------------- | ---------------------------------------- |
| $\exp(s_k(x))$   | $e^{s_k}$            | Экспонента score класса $k$              |
| $\sum_{j=1}^{K}$ | $\sum_j$             | Суммируем по всем классам                |
| $s_k(x)$         | $s_k$                | Score класса $k$ для объекта $x$         |
| $K$              | —                    | Количество классов, понятно из контекста |
| $\sigma(s(x))_k$ | —                    | Результат Softmax для класса $k$         |

То есть:

$$
\exp(s_k(x)) \equiv e^{s_k}
$$

и

$$
\sum_{j=1}^{K}\exp(s_j(x))
\equiv
\sum_j e^{s_j}
$$

Мы просто не указываем $x$ и границы суммы явно, потому что они уже понятны из контекста.

**Важно:** в книге $s_k(x)$ — это **score (logit)** класса $k$, а $\hat{p}_k$ — уже **вероятность** после применения Softmax.
