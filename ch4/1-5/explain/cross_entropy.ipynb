
# Cross Entropy — подробный разбор формул из книги

При обучении мы хотим, чтобы модель давала **высокую вероятность правильному классу** и низкую — остальным.

**Cross entropy** измеряет, насколько хорошо предсказанные вероятности совпадают с правильным классом.

> Чем ниже вероятность правильного класса, тем больше штраф.

Поэтому обучение старается **минимизировать Cross Entropy**.

---

## 1. Cost Function — Cross Entropy

Формула из книги:

$$
J(\Theta)
=
-\frac{1}{m}
\sum_{i=1}^{m}
\sum_{k=1}^{K}
y_k^{(i)}
\log\left(\hat{p}_k^{(i)}\right)
$$

Псевдокод:

```python
J = -mean(
    sum(
        y[i][k] * log(p_hat[i][k])
        for k in classes
    )
    for i in samples
)
```

### Что означает каждая часть

- $J(\Theta)$ — функция стоимости (**cost function**)
- $\Theta$ — все параметры модели
- $m$ — количество объектов
- $K$ — количество классов
- $\hat{p}_k^{(i)}$ — предсказанная вероятность того, что объект $i$ относится к классу $k$
- $y_k^{(i)}$ — правильный ответ для объекта $i$ и класса $k$

---

## 2. Что такое $y_k^{(i)}$

$$
y_k^{(i)} =
\begin{cases}
1, & \text{если объект относится к классу } k \\
0, & \text{иначе}
\end{cases}
$$

Псевдокод:

```python
y = 1 if sample_belongs_to_class else 0
```

То есть $y$ — это **не предсказание модели**.

Это правильный ответ, записанный как `0` или `1`.

Например, если правильный класс — Cat:

```text
Cat    → 1
Dog    → 0
Rabbit → 0
```

Это **one-hot encoding**.

---

## 3. Почему в формуле есть $y_k$

Рассмотрим:

$$
y_k^{(i)}
\log\left(\hat{p}_k^{(i)}\right)
$$

Псевдокод:

```python
term = y[i][k] * log(p_hat[i][k])
```

Если класс неправильный:

$$
y_k^{(i)}=0
$$

то:

$$
0\cdot\log\left(\hat{p}_k^{(i)}\right)=0
$$

Псевдокод:

```python
term = 0 * log(predicted_probability)
# term = 0
```

Этот класс не влияет на loss.

Если класс правильный:

$$
y_k^{(i)}=1
$$

то:

$$
1\cdot\log\left(\hat{p}_k^{(i)}\right)
=
\log\left(\hat{p}_k^{(i)}\right)
$$

Псевдокод:

```python
term = 1 * log(target_probability)
# term = log(target_probability)
```

Поэтому для one-hot target в итоге нас фактически интересует **вероятность правильного класса**.

---

## 4. Пример

Модель получила:

```text
Cat    → 0.725
Dog    → 0.267
Rabbit → 0.008
```

Правильный ответ:

```text
Cat    → 1
Dog    → 0
Rabbit → 0
```

Тогда:

$$
L
=
-\left(
1\log(0.725)
+
0\log(0.267)
+
0\log(0.008)
\right)
$$

Псевдокод:

```python
L = -(
    1 * log(0.725)
    + 0 * log(0.267)
    + 0 * log(0.008)
)
```

Получаем:

$$
L=-\log(0.725)\approx0.322
$$

Псевдокод:

```python
L = -log(0.725)
# ≈ 0.322
```

Loss небольшой, потому что модель дала правильному классу высокую вероятность.

---

## 5. Если модель ошиблась

Пусть правильный класс всё тот же Cat:

```text
Cat    → 0.01
Dog    → 0.80
Rabbit → 0.19
```

Тогда:

$$
L=-\log(0.01)\approx4.605
$$

Псевдокод:

```python
L = -log(0.01)
# ≈ 4.605
```

Loss намного больше.

Получается:

```text
P(correct) = 0.9
→ loss ≈ 0.105

P(correct) = 0.5
→ loss ≈ 0.693

P(correct) = 0.1
→ loss ≈ 2.303

P(correct) = 0.01
→ loss ≈ 4.605
```

**Чем сильнее модель уверена в неправильном ответе, тем сильнее её штрафуют.**

---

## 6. Почему используется $-\log$

$$
-\log(p)
$$

Псевдокод:

```python
loss = -log(p)
```

Поведение функции:

```text
p → 1
loss → 0

p → 0
loss → ∞
```

Поэтому:

> **высокая вероятность правильного класса → маленький штраф**

> **низкая вероятность правильного класса → большой штраф**

---

## 7. Что делает вся формула

$$
J(\Theta)
=
-\frac{1}{m}
\sum_{i=1}^{m}
\sum_{k=1}^{K}
y_k^{(i)}
\log\left(\hat{p}_k^{(i)}\right)
$$

Псевдокод:

```python
total_loss = 0

for sample in dataset:
    for class_k in classes:
        total_loss += (
            -y[sample][class_k]
            * log(p_hat[sample][class_k])
        )

J = total_loss / m
```

Можно читать буквально:

```text
для каждого объекта
    найти правильный класс
    посмотреть вероятность этого класса
    вычислить -log(вероятности)

затем
    сложить loss всех объектов
    разделить на количество объектов
```

---

## 8. Почему это называется Cross Entropy

В общем виде Cross Entropy определяется так:

$$
H(p,q)
=
-\sum_x p(x)\log q(x)
$$

Псевдокод:

```python
H = -sum(
    p[x] * log(q[x])
    for x in outcomes
)
```

Здесь:

- $p$ — настоящее распределение
- $q$ — предсказанное распределение

В нашей задаче:

```text
p = правильное распределение
q = предсказанные моделью вероятности
```

Например:

```text
Правильный ответ:

Cat    → 1
Dog    → 0
Rabbit → 0
```

Предсказание:

```text
Cat    → 0.725
Dog    → 0.267
Rabbit → 0.008
```

Cross entropy сравнивает эти два распределения.

Для one-hot target формула фактически сводится к:

$$
-\log\left(\hat{p}_{target}\right)
$$

Псевдокод:

```python
loss = -log(target_probability)
```

То есть в классификации нас фактически интересует:

> **какую вероятность модель дала правильному классу.**

> **Важно:** если используется натуральный логарифм $\ln$, значение измеряется в **натах**. Если использовать $\log_2$, значение можно интерпретировать в **битах**.

---

## 9. Связь с бинарной логистической регрессией

Книга отмечает, что при двух классах:

$$
K=2
$$

Псевдокод:

```python
K = 2
```

эта функция стоимости эквивалентна log loss из логистической регрессии:

$$
J(\theta)
=
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y^{(i)}\log\left(\hat{p}^{(i)}\right)
+
(1-y^{(i)})
\log\left(1-\hat{p}^{(i)}\right)
\right]
$$

Псевдокод:

```python
J = -mean(
    y * log(p_hat)
    + (1 - y) * log(1 - p_hat)
)
```

То есть **binary log loss — это частный случай Cross Entropy**.

---


````python
import numpy as np


# Это правая часть формулы с (sum _ yk _ pk)
# вся цель данного приема в том чтобы понять на сколько мы ошибаемся для нашего таргета
targets = np.array([
    [1, 0, 0],
    [1, 0, 0],
])

probs = np.array([
    [0.725, 0.267, 0.008],   # объект 1: модель почти уверена, что cat
    [0.010, 0.800, 0.190],   # объект 2: модель почти уверена, что dog
])

log_probs = np.log(probs + 1e-12)
print(log_probs,"\n")

# умножение таргетов на пробы
print(targets * log_probs, "\n")

# Суммируем ошибки!
per_sample_loss = -np.sum(targets * log_probs, axis=1)
print(per_sample_loss)
````

````python
import numpy as np

# 1) ПРОСТОЙ ПРИМЕР НА 3 КЛАССАХ (cat / dog / rabbit)

# Вероятности, которые "выдала" модель (после softmax)
# Строки = объекты, столбцы = [cat, dog, rabbit]
probs = np.array([
    [0.725, 0.267, 0.008],   # объект 1: модель почти уверена, что cat
    [0.010, 0.800, 0.190],   # объект 2: модель почти уверена, что dog
])

# Правильные классы (one-hot: где 1 — там верный класс)
# Пусть оба объекта на самом деле CAT
targets = np.array([
    [1, 0, 0],
    [1, 0, 0],
])

# Формула: loss_i = -sum_k  y_k * log(p_k)
#          loss   = (1/m) * sum_i loss_i
#
# Заметь: из-за one-hot y_k != 0 только для правильного класса,
# поэтому на самом деле берём только log(p_правильный).

log_probs = np.log(probs + 1e-12)          # +1e-12 защита от log(0)
per_sample_loss = -np.sum(targets * log_probs, axis=1)
loss = per_sample_loss.mean()

print("loss по образцу 1:", round(per_sample_loss[0], 4))   # -log(0.725) ≈ 0.3216
print("loss по образцу 2:", round(per_sample_loss[1], 4))   # -log(0.01)  ≈ 4.6052
print("СРЕДНИЙ cross entropy:", round(loss, 4))            # ≈ 2.4634

````

````python
# 2) ЭТО ЖЕ, НО "КАК В НАСТОЯЩЕМ ML" (softmax + one-hot)

def softmax(z):
    e = np.exp(z - z.max(axis=1, keepdims=True))  # трюк для стабильности
    return e / e.sum(axis=1, keepdims=True)

# "сырые" очки модели (до softmax)
raw = np.array([
    [2.0, -1.0, -3.0],   # cat сильно впереди -> почти правильный
    [-3.0, 2.0, -1.0],   # dog впереди -> почти неверный (правильный cat)
])

p = softmax(raw)
y = np.array([[1, 0, 0],
              [1, 0, 0]])

ce = -np.mean(np.sum(y * np.log(p + 1e-12), axis=1))
print("\nCross entropy через softmax:", round(ce, 4))
````

````python
# 3) СВЯЗЬ С БИТАМИ (бинарный случай, log2)
# Реальность: 80% ясно, 20% дождь
p_real = np.array([0.8, 0.2])

def bits_cross_entropy(p_true, p_model):
    return -np.sum(p_true * np.log2(p_model))

# Модель А: "идеальная", совпадает с реальностью
model_A = np.array([0.8, 0.2])
# Модель Б: "безыдейная", говорит 50/50
model_B = np.array([0.5, 0.5])

print("\nБиты на исход (модель А, идеальная):", round(bits_cross_entropy(p_real, model_A), 4))
print("Биты на исход (модель Б, 50/50)    :", round(bits_cross_entropy(p_real, model_B), 4))
print("Лишние биты = KL-дивергенция      :",
      round(bits_cross_entropy(p_real, model_B) - bits_cross_entropy(p_real, model_A), 4))
````
