# Gradient — подробный разбор формул из книги

После Cross Entropy мы уже умеем измерять ошибку модели:

$$
J(\Theta)
$$

Теперь возникает следующий вопрос:

> **Как изменить параметры модели, чтобы $J(\Theta)$ уменьшился?**

Для этого используется **Gradient**.

---

## 1. Формула Gradient

Формула из книги:

$$
\nabla_{\boldsymbol{\theta}^{(k)}} J(\Theta)
=
\frac{1}{m}
\sum_{i=1}^{m}
\left(
\hat{p}_k^{(i)} - y_k^{(i)}
\right)
x^{(i)}
$$

Псевдокод:

```python
gradient_k = mean(
    (p_hat_k - y_k) * x
)
```

Здесь:

- $\nabla_{\boldsymbol{\theta}^{(k)}}J(\Theta)$ — градиент функции стоимости по параметрам класса $k$
- $m$ — количество обучающих объектов
- $\hat{p}_k^{(i)}$ — предсказанная вероятность класса $k$ для объекта $i$
- $y_k^{(i)}$ — правильный ответ для класса $k$: `1` или `0`
- $x^{(i)}$ — признаки объекта $i$

---

## 2. Что вообще показывает Gradient?

Градиент отвечает на вопрос:

> **Если немного изменить параметры $\theta$, как изменится loss?**

Сам Gradient указывает направление, в котором функция стоимости **растёт** быстрее всего.

Поэтому если хотим уменьшить loss, идём в противоположную сторону.

```text
Gradient
    ↓
направление роста loss

Gradient Descent
    ↓
идём в противоположную сторону
    ↓
loss уменьшается
```

---

## 3. Самая важная часть — $\hat{p}-y$

В формуле есть:

$$
\hat{p}_k^{(i)} - y_k^{(i)}
$$

Псевдокод:

```python
error = prediction - target
```

Можно интуитивно читать как:

> **предсказание − правильный ответ**

Например:

```text
Правильный класс = Cat

y:

Cat    → 1
Dog    → 0
Rabbit → 0
```

Модель предсказала:

```text
p:

Cat    → 0.725
Dog    → 0.267
Rabbit → 0.008
```

Тогда:

### Cat

$$
0.725 - 1 = -0.275
$$

```python
error_cat = 0.725 - 1
# -0.275
```

### Dog

$$
0.267 - 0 = 0.267
$$

```python
error_dog = 0.267 - 0
# 0.267
```

### Rabbit

$$
0.008 - 0 = 0.008
$$

```python
error_rabbit = 0.008 - 0
# 0.008
```

Получаем:

```text
Cat    → -0.275
Dog    →  0.267
Rabbit →  0.008
```

---

## 4. Что означает знак?

Здесь важно помнить, что сам Gradient указывает направление **роста loss**, а обновление параметров выполняется с минусом.

### Для правильного класса

$$
\hat{p}-y < 0
$$

Например:

$$
0.725-1=-0.275
$$

Отрицательный вклад говорит, что для правильного класса модель дала вероятность меньше `1`.

При Gradient Descent обновление с минусом направляет параметры так, чтобы **повысить score этого класса** для похожих объектов.

---

### Для неправильного класса

$$
\hat{p}-y > 0
$$

Например:

$$
0.267-0=0.267
$$

Положительный вклад означает, что модель дала неправильному классу ненулевую вероятность.

Gradient Descent направляет обновление параметров так, чтобы **снизить score этого класса** для похожих объектов.

---

## 5. Важное уточнение

Не стоит понимать знак как:

> `-` всегда просто «увеличить параметр»  
> `+` всегда просто «уменьшить параметр»

Потому что в формуле есть ещё признаки:

$$
\left(\hat{p}_k^{(i)}-y_k^{(i)}\right)x^{(i)}
$$

А признаки $x$ могут быть как положительными, так и отрицательными.

Правильнее говорить:

> **Gradient определяет, как изменить каждый параметр с учётом ошибки и значения соответствующего признака.**

---

## 6. Зачем умножение на $x$

Рассмотрим:

$$
\left(
\hat{p}_k^{(i)}-y_k^{(i)}
\right)x^{(i)}
$$

Псевдокод:

```python
gradient_contribution = error * features
```

Здесь:

$$
x^{(i)}
$$

— признаки конкретного объекта.

То есть вклад объекта зависит от двух вещей:

```text
насколько модель ошиблась
        ×
какие признаки были у объекта
```

Например:

```text
error = -0.3

features = [3, 2]
```

Тогда:

```text
error * features

= -0.3 * [3, 2]

= [-0.9, -0.6]
```

Каждый признак даёт свой вклад в изменение соответствующего параметра.

---

## 7. Почему делим на $m$

Формула:

$$
\frac{1}{m}
\sum_{i=1}^{m}
\left(
\hat{p}_k^{(i)}-y_k^{(i)}
\right)x^{(i)}
$$

сначала складывает вклад всех объектов:

$$
\sum_{i=1}^{m}
\left(
\hat{p}_k^{(i)}-y_k^{(i)}
\right)x^{(i)}
$$

а затем делит на количество объектов:

$$
\frac{1}{m}
$$

То есть мы получаем **средний градиент по всей обучающей выборке**.

Псевдокод:

```python
gradient = mean(
    [
        (p_hat[i][k] - y[i][k]) * x[i]
        for i in samples
    ]
)
```

---

## 8. Почему градиент отдельный для каждого класса?

У каждого класса свои параметры:

$$
\boldsymbol{\theta}^{(1)},
\boldsymbol{\theta}^{(2)},
\dots,
\boldsymbol{\theta}^{(K)}
$$

Поэтому для каждого класса нужен свой градиент:

$$
\nabla_{\boldsymbol{\theta}^{(1)}}J(\Theta)
$$

$$
\nabla_{\boldsymbol{\theta}^{(2)}}J(\Theta)
$$

$$
\dots
$$

$$
\nabla_{\boldsymbol{\theta}^{(K)}}J(\Theta)
$$

Книга поэтому говорит:

> **Now you can compute the gradient vector for every class.**

То есть мы отдельно вычисляем направление обновления параметров для каждого класса.

---

## 9. Gradient Descent

После вычисления градиента нужно изменить параметры.

Формула:

$$
\boldsymbol{\theta}
\leftarrow
\boldsymbol{\theta}
-
\eta
\nabla_{\boldsymbol{\theta}}J(\Theta)
$$

Псевдокод:

```python
theta = theta - learning_rate * gradient
```

Здесь:

- $\eta$ — **learning rate**
- $\nabla_{\boldsymbol{\theta}}J(\Theta)$ — Gradient
- $\boldsymbol{\theta}$ — параметры модели

Почему стоит минус?

Потому что:

```text
Gradient
↓
направление роста loss

- Gradient
↓
направление уменьшения loss
```

---

## 10. Простая интуиция

Представь, что loss — это гора:

```text
        loss ↑
             /\
            /  \
           /    \
          /      \
         /        \
        ●
```

Gradient показывает, куда поверхность поднимается быстрее всего.

Мы хотим идти вниз:

```text
        loss ↑
             /\
            /  \
           /    \
      ● ←  /      \
         ↓
      минимум
```

Поэтому:

$$
\theta \leftarrow \theta-\eta\nabla_\theta J(\Theta)
$$

---

## 11. Вся последовательность обучения

```text
x
↓
Score
↓
Softmax
↓
вероятности
↓
Cross Entropy
↓
Loss
↓
Gradient
↓
Gradient Descent
↓
обновляем θ
↓
снова считаем Score
↓
снова Softmax
↓
...
```

Псевдокод:

```python
while not_converged:

    scores = compute_scores(X, theta)

    predictions = softmax(scores)

    loss = cross_entropy(y, predictions)

    gradient = compute_gradient(
        X,
        y,
        predictions
    )

    theta = theta - learning_rate * gradient
```

Цель:

$$
\boxed{
\min_{\Theta} J(\Theta)
}
$$

То есть найти такие параметры $\Theta$, при которых Cross Entropy будет минимальной.
