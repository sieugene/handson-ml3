## Argmax — формула из книги

В книге предсказание записано более подробно:

$$
\hat{y}
=
\arg\max_k \sigma(s(x))_k
=
\arg\max_k s_k(x)
=
\arg\max_k
\left((\boldsymbol{\theta}^{(k)})^\top x\right)
$$

```python
y_hat = argmax(softmax(s(x)))
      = argmax(s(x))
      = argmax(theta[k].T @ x)
```

> **Здесь $\sigma$ — обозначение функции Softmax.**
> Поэтому $\sigma(s(x))_k$ означает результат применения Softmax к scores для класса $k$, то есть вероятность этого класса.

Здесь одна и та же операция записана **тремя способами**.

### 1. Через вероятность после Softmax

$$
\hat{y}
=
\arg\max_k \sigma(s(x))_k
$$

```python
y_hat = argmax(softmax(scores))
```

То есть:

> **выбираем класс с максимальной вероятностью.**

Например:

```text
Кот      → 0.725
Собака   → 0.267
Кролик   → 0.008
```

Максимальная вероятность → **Кот**.

---

### 2. Через Score

$$
\arg\max_k \sigma(s(x))_k
=
\arg\max_k s_k(x)
$$

```python
argmax(softmax(scores)) == argmax(scores)
```

Softmax не меняет порядок значений:

```text
Score:

Кот      → 8
Собака   → 7
Кролик   → 3.5
```

После Softmax:

```text
Кот      → 0.725
Собака   → 0.267
Кролик   → 0.008
```

Поэтому максимальная вероятность всегда соответствует максимальному Score.

Значит, чтобы просто выбрать класс, **Softmax считать необязательно**:

$$
\hat{y} = \arg\max_k s_k(x)
$$

```python
y_hat = argmax(scores)
```

---

### 3. Через формулу Score

Мы уже знаем, что:

$$
s_k(x) = (\boldsymbol{\theta}^{(k)})^\top x
$$

```python
score_k = theta[k].T @ x
```

Поэтому:

$$
\arg\max_k s_k(x)
=
\arg\max_k
\left((\boldsymbol{\theta}^{(k)})^\top x\right)
$$

```python
y_hat = argmax([
    theta[0].T @ x,
    theta[1].T @ x,
    theta[2].T @ x
])
```

То есть можно сразу сказать:

> **Для каждого класса считаем $\theta^{(k)\top}x$ и выбираем класс, у которого результат максимальный.**
