```python
# Detect iris virginica
from sklearn.datasets import load_iris
import numpy as np
import matplotlib.pyplot as plt
iris = load_iris(as_frame=True)

X = iris.data.values
y = iris.target
def split_data(x, y):
    train_end = int(len(X) / 100 * 80)
    valid_start = train_end+1
    X_train = x[0:train_end]
    y_train = x[0:train_end]
    X_test = x[valid_start:len(X)]
    y_test = y[valid_start:len(X)]
    return X_train,y_train, X_test, y_test

X_train, y_train, X_test, y_test = split_data(X,y)
virginica_value_test = X[y][:1][0]
not_virginica_value_test = X[~y][:1][0]

```

```python
# Simple softmax
import math
import numpy as np

classes = iris.target_names


print("Формат данных X\n",X[0:3], "\n")
print("Формат данных Y\n",y[0:3], "\n")
print("Y к чему относится",classes[y[0:3]], "\n")
print("рассмотрим первый элемент, он - ",classes[y[0]])
print("возможные классы",classes)
print("-------")

# Получение оценки
def score_by_class(x):
    # Тут тета как и всегда просто рандомные начальные какие-то цифры
    # У нас не используется y чтобы проверить правда ли это или нет
    thetas = [
        # theta_1 'setosa' init
        [0.1, 0.2, 0.3, 0.4],
        # theta_2 'versicolor' init
        [0.3, 0.7, 0.5, 0.9],
        # theta_3 'virginica init
        [0.5, 0.1, 0.4, 0.7]
        ]
    scores = []
    for _,theta_k in enumerate(thetas):
        scores.append(theta_k @ x)
    return scores


score_for_x_1 = score_by_class(X[0])
argMax = max(score_for_x_1)
print("score for x_1",score_for_x_1)
print("argMax", argMax)
print("---------")

# Превращение в вероятности
def probability(scores):
    sum_exp = 0
    for score in scores:
        sum_exp+=math.exp(score)
    probs = []
    for score in scores:
        probs.append(math.exp(score) / sum_exp)
    return probs

probability_for_x_1 = probability(score_for_x_1)
print("probability for x_1 score",probability_for_x_1)
print("-------")

# Cross entropy
def hot_encode(values):
    onehot = {}
    for index,value in enumerate(values):
        # Сначала наполняем нулями
        onehot[value] = [0] * len(values)
        # Потом заменяем нужный элемент массива на 1
        onehot[value][index] = 1
    return onehot
one_hot = hot_encode(classes)
print("one hot", one_hot)
print("------")

# Σ  y_k · log(p̂_k)     (сумма по классам k = 1, 2, 3)
# one hot умножаем на prob
# 1 · log(p̂_setosa)  +  0 · log(p̂_versicolor)  +  0 · log(p̂_virginica) = log(p̂_setosa)

def crossEntropy(prob, correct_class):
    y_onehot = np.array(one_hot[correct_class])   # one-hot только для правильного
    return -np.sum(y_onehot * np.log(prob))       # минус и сумма по классам

loss_x1 = crossEntropy(probability_for_x_1, classes[y[0]])
print("loss for x_1", loss_x1) 

```
